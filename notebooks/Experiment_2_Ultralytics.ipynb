## 1. YOLOv10n
### 1.1 Baseline (Pretrained COCO)

```python
from ultralytics import YOLO

# Load pretrained COCO model
model = YOLO("yolov10n.pt")

# Evaluate pretrained baseline on real test set
metrics = model.val(data="mydata_capped.yaml", split="test")

print(f"\n=== YOLOv10n COCO Pretrained Baseline ===")
print(f"Precision:     {metrics.box.mp:.3f}")
print(f"Recall:        {metrics.box.mr:.3f}")
print(f"mAP@50:        {metrics.box.map50:.3f}")
print(f"mAP@50-95:     {metrics.box.map:.3f}")
```

### 1.2 Train on Real CCTV Data

```python
from ultralytics import YOLO

model = YOLO("yolov10n.pt")

model.train(
    data="mydata_capped.yaml",
    epochs=40,
    freeze=0,
    lr0=2e-4,
    imgsz=832,
    name="yolov10n_real_e40"
)

metrics = model.val(
    data="mydata_capped.yaml",
    split="test",
    plots=True,
    save_json=True
)

print("\n===== YOLOv10n FINAL TEST METRICS =====")
print(f"mAP@0.50:        {metrics.box.map50:.4f}")
print(f"mAP@0.50:0.95:   {metrics.box.map:.4f}")
print(f"Precision:       {metrics.box.mp:.4f}")
print(f"Recall:          {metrics.box.mr:.4f}")
print("========================================\n")
```
